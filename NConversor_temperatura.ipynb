El programa que convierte grados Celsius a Fahrenheit y al revés.     
Con validaciones para que no se rompa si alguien, el usuario se equivoca.

## Fórmulas que usamos

- Para pasar de **Celsius a Fahrenheit**:
  `F = C * (9/5) + 32`

- Para pasar de **Fahrenheit a Celsius**:
  `C = (F - 32) * (5/9)`

def celsius_a_fahrenheit(celsius):
   
    # Fórmula, multiplico por 9/5 y le sumo 32
    return celsius * (9 / 5) + 32


def fahrenheit_a_celsius(fahrenheit):
   
    # Aquí resto 32 y después multiplico por 5/9
    return (fahrenheit - 32) * (5 / 9)

## Validación de datos

También usamos una función que se encarga de pedir el número.

Si el usuario deja el campo vacío o pone letras, le avisa y le vuelve a pedir.
